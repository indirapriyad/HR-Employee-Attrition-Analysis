```python
import pandas as pd

data = {
    "Employee_ID": [101,102,103,104,105,106,107,108,109,110],
    "Age": [25,32,28,41,29,35,24,45,31,27],
    "Department": ["HR","IT","Sales","Finance","IT","HR","Sales","Finance","IT","HR"],
    "Salary": [28000,52000,35000,65000,48000,30000,32000,72000,50000,29000],
    "Years_at_Company": [1,5,3,10,4,2,1,12,6,2],
    "Job_Satisfaction": [3,4,2,5,4,3,2,5,4,3],
    "Attrition": ["Yes","No","Yes","No","No","Yes","Yes","No","No","Yes"]
}

df = pd.DataFrame(data)

df
```

```python
total_employees = len(df)

print("Total Employees:", total_employees)
```

```python
employees_left = (df["Attrition"] == "Yes").sum()

print("Employees who left:", employees_left)
```

```python
attrition_rate = (employees_left / total_employees) * 100

print("Attrition Rate:", attrition_rate, "%")
```

```python
average_salary = df["Salary"].mean()

print("Average Salary: ₹", average_salary)
```

```python
department_attrition = df[df["Attrition"] == "Yes"]["Department"].value_counts()

print(department_attrition)
```

```python
import matplotlib.pyplot as plt

# Attrition count
attrition_counts = df["Attrition"].value_counts()

plt.figure(figsize=(6,4))
attrition_counts.plot(kind="bar")
plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.show()
```

```python
department_attrition = df[df["Attrition"] == "Yes"]["Department"].value_counts()

plt.figure(figsize=(7,4))
department_attrition.plot(kind="bar")
plt.title("Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Employees Who Left")
plt.xticks(rotation=0)
plt.show()
```

```python
salary_by_department = df.groupby("Department")["Salary"].mean()

plt.figure(figsize=(7,4))
salary_by_department.plot(kind="bar")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.show()
```

```python
print("===== HR EMPLOYEE ANALYSIS =====")
print("Total Employees:", total_employees)
print("Employees Who Left:", employees_left)
print("Attrition Rate:", attrition_rate, "%")
print("Average Salary: ₹", average_salary)
print("\nAttrition by Department:")
print(department_attrition)
```

```python
import pandas as pd

df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.head()
```

```python
print("Total Employees:", len(df))
print("Total Columns:", len(df.columns))
```

```python
attrition_count = df["Attrition"].value_counts()

print(attrition_count)
```

```python
attrition_rate = (df["Attrition"] == "Yes").mean() * 100

print("Employee Attrition Rate:", round(attrition_rate, 2), "%")
```

```python
summary = {
    "Total Employees": len(df),
    "Employees Left": (df["Attrition"] == "Yes").sum(),
    "Employees Stayed": (df["Attrition"] == "No").sum(),
    "Attrition Rate (%)": round(attrition_rate, 2)
}

summary
```

```python
department_count = df["Department"].value_counts()

print(department_count)
```

```python
department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"]
)

department_attrition
```

```python
department_rate = (
    df.groupby("Department")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .round(2)
)

print(department_rate)
```

```python
highest_department = department_rate.idxmax()
highest_rate = department_rate.max()

print("Highest Attrition Department:", highest_department)
print("Attrition Rate:", highest_rate, "%")
```

```python
salary_attrition = df.groupby("Attrition")["MonthlyIncome"].mean()

print("Average Monthly Income:")
print(salary_attrition)
```

```python
satisfaction_attrition = pd.crosstab(
    df["JobSatisfaction"],
    df["Attrition"]
)

print(satisfaction_attrition)
```

```python
overtime_attrition = pd.crosstab(
    df["OverTime"],
    df["Attrition"]
)

print(overtime_attrition)
```

```python
overtime_rate = (
    df.groupby("OverTime")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .round(2)
)

print("Attrition Rate by Overtime:")
print(overtime_rate)
```

```python
print("====================================")
print("       HR ATTRITION DASHBOARD")
print("====================================")

print("Total Employees:", len(df))

employees_left = (df["Attrition"] == "Yes").sum()
print("Employees Who Left:", employees_left)

attrition_rate = (df["Attrition"] == "Yes").mean() * 100
print("Overall Attrition Rate:", round(attrition_rate, 2), "%")

average_income = df["MonthlyIncome"].mean()
print("Average Monthly Income: ₹", round(average_income, 2))

highest_department = department_rate.idxmax()
highest_department_rate = department_rate.max()

print("Highest Attrition Department:", highest_department)
print("Department Attrition Rate:", highest_department_rate, "%")
```

```python
import matplotlib.pyplot as plt

attrition_counts = df["Attrition"].value_counts()

plt.figure(figsize=(7,5))
attrition_counts.plot(kind="bar")

plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)

plt.show()
```

```python
department_rate.sort_values(ascending=False).plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

plt.show()
```

```python
overtime_rate.plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

plt.show()
```
